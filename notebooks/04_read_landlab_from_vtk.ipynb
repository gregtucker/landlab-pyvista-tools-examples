# Example of reading and plotting Landlab output in .vtk format

*(Greg Tucker, University of Colorado Boulder, September 2025)*

One of the nice attributes of PyVista is its close association with the Visualization Toolkit (VTK) format and data model. When a Landlab grid and its fields are [saved in a VTK-format file](https://landlab.csdms.io/tutorials/output/writing_legacy_vtk_files.html), it is easy to read them in as a PyVista mesh object.

The example below reads in a single frame of a Landlab landscape evolution model simulation, which happens to represent a large, circular island undergoing tectonic rifting, erosion, and sedimentation.

```python
import numpy as np
import pyvista as pv
```

Use the PyVista `read()` function to read the VTK contents into a PyVista mesh:

```python
llvtk = pv.read('rift-island-save141.vtk')
llvtk
```

Use the mesh's `plot()` method to display it:

```python
llvtk.plot()
```

One drawback is that the lack of vertical exaggeration makes it difficult to see the topography and bathymetry. To overcome that limitation, create a PyVista `Plotter` object, add the mesh to it, and set its vertical scale to 20x exaggeration:

```python
pl = pv.Plotter()
pl.add_mesh(llvtk)
pl.set_scale(zscale=20.0)
pl.show()
```

In this example, because the quantity *cumulative deposit thickness* could be either positive (indicating deposits) or negative (indicating erosion), it's useful to use a diverging color scale with zero erosion or deposition in the middle of the scale. Here's one way to do this:

```python
pl = pv.Plotter()
max_dep = np.amax(np.abs(llvtk["cumulative_deposit_thickness"]))
pl.add_mesh(llvtk, cmap="bwr", clim=[-max_dep, max_dep])
pl.set_scale(zscale=20.0)
pl.show()
```

We might also want to color the terrain by a different quantity. The `array_names` property contains a list of scalar fields:

```python
llvtk.array_names
```

These fields (arrays) are accessed using dictionary-like syntax. For example:

```python
print("The mean slope gradient is", np.mean(llvtk["topographic__steepest_slope"]))
```

The example below colors the topography by slope gradient:

```python
llvtk.set_active_scalars("topographic__steepest_slope")
pl = pv.Plotter()
pl.add_mesh(llvtk, cmap="hot")
pl.set_scale(zscale=20.0)
pl.show()
```
